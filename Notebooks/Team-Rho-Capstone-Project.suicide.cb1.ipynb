```python
"""
Google Trends Data Collection for Team Rho Capstone Project

Purpose:
Create state-year Google Trends predictors (2013-2023) for forecasting
future state-level suicide related inpatient psychiatric admissions.

"""

from pytrends.request import TrendReq
import pandas as pd
import time
import os
import random


# connect w/ google trends API, collect predictor search data

pytrends = TrendReq(
    hl="en-US",
    tz=360
)

# search terms/predictors

SEARCH_TERMS = [
    "suicidal thoughts",
    "suicide hotline",
    "self harm",
    "mental health crisis",
    "suicide prevention",
    "crisis hotline",
    "psychiatric hospital",
    "depression help"
]


# states

STATES = {
    "AL":"Alabama",
    "AK":"Alaska",
    "AZ":"Arizona",
    "AR":"Arkansas",
    "CA":"California",
    "CO":"Colorado",
    "CT":"Connecticut",
    "DE":"Delaware",
    "FL":"Florida",
    "GA":"Georgia",
    "HI":"Hawaii",
    "ID":"Idaho",
    "IL":"Illinois",
    "IN":"Indiana",
    "IA":"Iowa",
    "KS":"Kansas",
    "KY":"Kentucky",
    "LA":"Louisiana",
    "ME":"Maine",
    "MD":"Maryland",
    "MA":"Massachusetts",
    "MI":"Michigan",
    "MN":"Minnesota",
    "MS":"Mississippi",
    "MO":"Missouri",
    "MT":"Montana",
    "NE":"Nebraska",
    "NV":"Nevada",
    "NH":"New Hampshire",
    "NJ":"New Jersey",
    "NM":"New Mexico",
    "NY":"New York",
    "NC":"North Carolina",
    "ND":"North Dakota",
    "OH":"Ohio",
    "OK":"Oklahoma",
    "OR":"Oregon",
    "PA":"Pennsylvania",
    "RI":"Rhode Island",
    "SC":"South Carolina",
    "SD":"South Dakota",
    "TN":"Tennessee",
    "TX":"Texas",
    "UT":"Utah",
    "VT":"Vermont",
    "VA":"Virginia",
    "WA":"Washington",
    "WV":"West Virginia",
    "WI":"Wisconsin",
    "WY":"Wyoming",
    "DC":"District of Columbia"
}


# create files

RAW_FILE = "google_trends_raw_progress.csv"

FINAL_FILE = "google_trends_state_year_2013_2023.csv"

FAILED_FILE = "google_trends_failed_queries.csv"


TIMEFRAME = "2013-01-01 2023-12-31"

MAX_RETRIES = 5


# load google trends data

if os.path.exists(RAW_FILE):

    raw = pd.read_csv(RAW_FILE)

    records = [raw]

    completed = set(
        zip(
            raw["State"],
            raw["Search_Term"]
        )
    )

else:

    records = []

    completed = set()


failed_queries = []


# collect google trends data, handle API fail

def get_google_trends(term, geo):

    for attempt in range(MAX_RETRIES):

        try:

            pytrends.build_payload(
                kw_list=[term],
                timeframe=TIMEFRAME,
                geo=geo
            )

            data = pytrends.interest_over_time()

            if not data.empty:

                return data


        except Exception as e:

            wait = 10 * (2 ** attempt)

            print(
                f"Retry {attempt+1}/{MAX_RETRIES}; waiting {wait}s"
            )

            time.sleep(wait)


    return None



# download predictor data

for state_abbrev, state_name in STATES.items():

    geo = f"US-{state_abbrev}"

    print(
        f"\nProcessing {state_name}"
    )


    for term in SEARCH_TERMS:


        if (state_name, term) in completed:

            print(
                f"Skipping {state_name} - {term}"
            )

            continue


        trends = get_google_trends(
            term,
            geo
        )


        if trends is None:

            failed_queries.append(
                {
                    "State": state_name,
                    "Term": term
                }
            )

            continue


        trends = trends.reset_index()

        trends["Year"] = (
            trends["date"]
            .dt.year
        )


        annual = (
            trends
            .groupby("Year")[term]
            .mean()
            .reset_index()
        )


        annual["State"] = state_name

        annual["Search_Term"] = term


        records.append(
            annual
        )


        # Save checkpoint immediately
        checkpoint = pd.concat(
            records,
            ignore_index=True
        )


        checkpoint.to_csv(
            RAW_FILE,
            index=False
        )


        time.sleep(
            random.uniform(5,10)
        )


# convert long-format data into one row per state-year

raw = pd.concat(
    records,
    ignore_index=True
)


panel = (
    raw
    .pivot_table(
        index=[
            "State",
            "Year"
        ],
        columns="Search_Term",
        values=raw.columns[1],
        aggfunc="mean"
    )
    .reset_index()
)


panel.columns = [
    str(c)
    .replace(" ","_")
    .replace("-","_")
    .lower()
    for c in panel.columns
]


# create one year timelag variables to test hypothesis

panel = panel.sort_values(
    [
        "state",
        "year"
    ]
)


predictors = [
    c for c in panel.columns
    if c not in [
        "state",
        "year"
    ]
]


for col in predictors:

    panel[col + "_lag1"] = (
        panel
        .groupby("state")[col]
        .shift(1)
    )


# save data set

panel.to_csv(
    FINAL_FILE,
    index=False
)


if failed_queries:

    pd.DataFrame(
        failed_queries
    ).to_csv(
        FAILED_FILE,
        index=False
    )


print("COMPLETE")
print(panel.head())
```

```python
# build State Population Dataset (2013-2023), use following
#   nst-est2020.xlsx
#   NST-EST2025-POP.xlsx


import pandas as pd
from pathlib import Path

# file locations

DATA_DIR = Path("/Users/caseybrookshier/Desktop/Data Science job attainment planning/DS Portfolio github friendly/Team-Rho-Capstone-Project.cb/data/raw")

OLD_FILE = DATA_DIR / "nst-est2020.xlsx"
NEW_FILE = DATA_DIR / "NST-EST2025-POP.xlsx"

OUTPUT_FILE = DATA_DIR / "uscb_state_population_2013_2023.csv"


# read worksheet

old_raw = pd.read_excel(OLD_FILE, header=None)
new_raw = pd.read_excel(NEW_FILE, header=None)

# locate first state (.Alabama)


old_start = old_raw[old_raw.iloc[:,0].astype(str).str.startswith(".Alabama")].index[0]
new_start = new_raw[new_raw.iloc[:,0].astype(str).str.startswith(".Alabama")].index[0]



# build OLD (2013-2020) state pop.


old_years = [
    "State","2010","Base","2010","2011","2012",
    "2013","2014","2015","2016","2017",
    "2018","2019","2020_Apr","2020"
]

old = old_raw.iloc[old_start:].copy()
old.columns = old_years

old = old[
    ["State","2013","2014","2015","2016",
     "2017","2018","2019","2020"]
]

# build NEW (2021-2023)

new_years = [
    "State","Base","2020","2021",
    "2022","2023","2024","2025"
]

new = new_raw.iloc[new_start:].copy()
new.columns = new_years

new = new[
    ["State","2021","2022","2023"]
]

# merge

wide = old.merge(new,on="State",how="left")

# remove leading dots from state names
wide["State"] = (
    wide["State"]
    .astype(str)
    .str.replace(".","",regex=False)
    .str.strip()
)

# keep only states + DC
exclude = {
    "United States",
    "Northeast",
    "Midwest",
    "South",
    "West",
    "Puerto Rico"
}

wide = wide[~wide["State"].isin(exclude)]

# convert to long format

population = wide.melt(
    id_vars="State",
    var_name="Year",
    value_name="state_population"
)

population["Year"] = population["Year"].astype(int)
population["state_population"] = (
    pd.to_numeric(population["state_population"], errors="coerce")
    .astype("Int64")
)

population = (
    population
    .sort_values(["State","Year"])
    .reset_index(drop=True)
)

population.to_csv(OUTPUT_FILE,index=False)

print("Saved:", OUTPUT_FILE)
print("Rows:", len(population))
print(population.head(20))
```

```python

# HCUP MENTAL HEALTH ADMISSIONS CLEANING PIPELINE
# Input:DownloadTable_StatePayerIP_2026-03-20.xls
# Output:hcup_state_year_mental_health_admissions_2013_2023.csv
#   create accurate State x Year mental health inpatient
#   admissions table for predictive modeling.


import pandas as pd
import numpy as np
from pathlib import Path

# paths


RAW = Path(
    "/Users/caseybrookshier/Desktop/Data Science job attainment planning/"
    "DS Portfolio github friendly/Team-Rho-Capstone-Project.cb/data/raw"
)


HCUP_FILE = RAW / "DownloadTable_StatePayerIP_2026-03-20.xls"


OUTPUT = RAW / (
    "hcup_state_year_mental_health_admissions_2013_2023.csv"
)

# load HCUP raw sheet


raw = pd.read_excel(
    HCUP_FILE,
    sheet_name="Data",
    header=None
)


print("Raw HCUP shape:")
print(raw.shape)

# find header row

header_row = None


for i in range(len(raw)):

    row = raw.iloc[i].astype(str).tolist()

    if (
        "State" in row
        and "Hospitalization Type" in row
        and "Expected Payer" in row
    ):
        header_row = i
        break



if header_row is None:
    raise ValueError(
        "HCUP header row not found."
    )


print("\nUsing header row:")
print(header_row)

# reload with correct header

hcup = pd.read_excel(
    HCUP_FILE,
    sheet_name="Data",
    header=header_row
)

hcup.columns = (
    hcup.columns
    .astype(str)
    .str.strip()
)

# identify quarterly columns

quarter_cols = [
    c for c in hcup.columns
    if (
        isinstance(c, str)
        and len(c) >= 7
        and c[:4].isdigit()
        and "Q" in c
    )
]


print("\nQuarter columns detected:")
print(len(quarter_cols))

# numeric conversion

for c in quarter_cols:

    hcup[c] = (
        hcup[c]
        .astype(str)
        .str.replace(",", "", regex=False)
        .replace(
            ["nan", "None", ""],
            np.nan
        )
    )


    hcup[c] = pd.to_numeric(
        hcup[c],
        errors="coerce"
    )

# keep 2013-2023 only

analysis_quarters = [
    c for c in quarter_cols
    if any(
        str(y) in c
        for y in range(2013,2024)
    )
]


hcup = hcup[
    [
        "State",
        "Hospitalization Type",
        "Expected Payer"
    ]
    +
    analysis_quarters
]

# mental health only

mh = hcup[
    hcup["Hospitalization Type"]
    ==
    "Mental Health/Substance Use"
].copy()



print("\nMental health rows:")
print(mh.shape)

# remove HCUP calculated rows


summary_patterns = (
    "Combined|"
    "Sum|"
    "All expected"
)


mh = mh[
    ~mh["Expected Payer"]
    .str.contains(
        summary_patterns,
        case=False,
        regex=True,
        na=False
    )
]

# keep only independent payer groups


valid_payers = [
    "Medicare, age 65+",
    "Medicaid, age 19-64",
    "Private, age 19-64",
    "Self-Pay/No Charge, age 19-64"
]


mh = mh[
    mh["Expected Payer"]
    .isin(valid_payers)
]



print("\nRemaining payer categories:")
print(
    mh["Expected Payer"]
    .unique()
)



print("\nRows after payer filtering:")
print(mh.shape)

# aggregate payer categories

mh_state_quarter = (
    mh
    .groupby("State")[analysis_quarters]
    .sum(min_count=1)
    .reset_index()
)

# quarterly -> Annual


annual_records = []


for year in range(2013,2024):


    cols = [
        c for c in analysis_quarters
        if str(year) in c
    ]


    temp = mh_state_quarter[
        ["State"] + cols
    ].copy()



    temp["Year"] = year



    temp[
        "mental_health_admissions"
    ] = (
        temp[cols]
        .sum(
            axis=1,
            min_count=1
        )
    )



    annual_records.append(
        temp[
            [
                "State",
                "Year",
                "mental_health_admissions"
            ]
        ]
    )



hcup_clean = pd.concat(
    annual_records,
    ignore_index=True
)

# remove impossible outputs

hcup_clean = hcup_clean[
    hcup_clean["mental_health_admissions"]
    >= 0
]

# validation

print("\n==============================")
print("FINAL VALIDATION")
print("==============================")


print("\nShape:")
print(
    hcup_clean.shape
)


print("\nStates:")
print(
    hcup_clean["State"]
    .nunique()
)


print("\nYears:")
print(
    sorted(
        hcup_clean["Year"]
        .unique()
    )
)



print("\nDuplicate State-Year rows:")
print(
    hcup_clean
    .duplicated(
        ["State","Year"]
    )
    .sum()
)



print("\nAdmission distribution:")
print(
    hcup_clean[
        "mental_health_admissions"
    ]
    .describe()
)



print("\nTop state-year values:")
print(
    hcup_clean
    .sort_values(
        "mental_health_admissions",
        ascending=False
    )
    .head(15)
)

# notebook display

display(
    hcup_clean
    .sort_values(
        ["State","Year"]
    )
    .head(50)
)

#save

hcup_clean.to_csv(
    OUTPUT,
    index=False
)


print("\nSaved:")
print(OUTPUT)
```

```python
# TEAM RHO FINAL MERGE PIPELINE

# Inputs:
#  google_trends_raw_progress.csv
#  hcup_state_year_mental_health_admissions_2013_2023.csv
#  uscb_state_population_2013_2023.csv

# Output: team_rho_state_year_prediction_dataset.csv


import pandas as pd
from pathlib import Path

# Paths

RAW = Path(
    "/Users/caseybrookshier/Desktop/Data Science job attainment planning/"
    "DS Portfolio github friendly/Team-Rho-Capstone-Project.cb/data/raw"
)

GT_FILE = RAW / "google_trends_raw_progress.csv"

HCUP_FILE = RAW / (
    "hcup_state_year_mental_health_admissions_2013_2023.csv"
)

CENSUS_FILE = RAW / (
    "uscb_state_population_2013_2023.csv"
)

OUTPUT = RAW / (
    "team_rho_state_year_prediction_dataset.csv"
)

# GOOGLE TRENDS

google = pd.read_csv(GT_FILE)

google["State"] = google["State"].astype(str).str.strip()
google["Year"] = google["Year"].astype(int)

search_terms = [
    "suicidal thoughts",
    "suicide hotline",
    "self harm",
    "mental health crisis",
    "suicide prevention",
    "crisis hotline",
    "psychiatric hospital",
    "depression help"
]

for term in search_terms:
    google[term] = pd.to_numeric(
        google[term],
        errors="coerce"
    ).fillna(0)

google_year = (
    google
    .groupby(["State", "Year"], as_index=False)[search_terms]
    .sum()
)

# HCUP


hcup = pd.read_csv(HCUP_FILE)

hcup["State"] = hcup["State"].astype(str).str.strip()
hcup["Year"] = hcup["Year"].astype(int)

print("\nHCUP columns:")
print(hcup.columns.tolist())

required = [
    "State",
    "Year",
    "mental_health_admissions"
]

missing = [
    c for c in required
    if c not in hcup.columns
]

if missing:
    raise ValueError(
        f"HCUP file missing required columns: {missing}"
    )

# CENSUS

census = pd.read_csv(CENSUS_FILE)

census = census[
    census["State"].notna()
].copy()

census = census[
    census["state_population"].notna()
]

census["State"] = (
    census["State"]
    .astype(str)
    .str.strip()
)

census["Year"] = (
    census["Year"]
    .astype(int)
)

census = census[
    census["Year"].between(2013, 2023)
]

#MERGE

dataset = (
    google_year
    .merge(
        hcup,
        on=["State", "Year"],
        how="inner"
    )
    .merge(
        census,
        on=["State", "Year"],
        how="inner"
    )
)

print("\nMerged shape:")
print(dataset.shape)



# REMOVE HCUP MISSING VALUES

# note HCUP zeros represent unavailable reporting.

dataset = dataset[
    dataset["mental_health_admissions"] > 0
].copy()

#  ADMISSION RATE

dataset["admission_rate_per_100k"] = (
    dataset["mental_health_admissions"]
    /
    dataset["state_population"]
    * 100000
)

# CREATE LAGGED GOOGLE VARIABLES

dataset = dataset.sort_values(
    ["State", "Year"]
)

for term in search_terms:

    dataset[f"{term}_lag1"] = (
        dataset
        .groupby("State")[term]
        .shift(1)
    )

dataset = dataset[
    dataset["Year"] >= 2014
].copy()


#  KEEP MODEL VARIABLES


final_columns = [
    "State",
    "Year",
    "state_population",
    "mental_health_admissions",
    "admission_rate_per_100k"
]

final_columns.extend(
    [
        f"{term}_lag1"
        for term in search_terms
    ]
)

dataset = dataset[
    final_columns
]

# FINAL VALIDATION

print("\n" + "=" * 70)
print("TEAM RHO FINAL DATASET")
print("=" * 70)

print("\nShape:")
print(dataset.shape)

print("\nStates:")
print(dataset["State"].nunique())

print("\nYears:")
print(sorted(dataset["Year"].unique()))

print("\nDuplicate State-Year rows:")
print(
    dataset.duplicated(
        ["State", "Year"]
    ).sum()
)

print("\nMissing values:")
print(dataset.isna().sum())

print("\nAdmission rate summary:")
print(
    dataset["admission_rate_per_100k"]
    .describe()
)

print("\nOutcome summary:")
print(
    dataset["mental_health_admissions"]
    .describe()
)

print("\nFirst 20 rows:")
display(
    dataset.head(20)
)

# SAVE

dataset.to_csv(
    OUTPUT,
    index=False
)

print("\nSaved to:")
print(OUTPUT)
```
